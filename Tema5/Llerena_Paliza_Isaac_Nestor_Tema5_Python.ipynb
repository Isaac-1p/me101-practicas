```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("estudiantes_limpio.csv")

media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std()
var_poblacional = np.var(df["nota"])
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1

print("Media:", media)
print("Mediana:", mediana)
print("Desv. estandar muestral (pandas):", std_muestral)
print("Varianza poblacional (numpy):", var_poblacional)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

```python
skew_pandas = df["nota"].skew()
kurt_pandas = df["nota"].kurt()
skew_scipy = stats.skew(df["nota"])
kurt_scipy = stats.kurtosis(df["nota"])

print("Skewness pandas:", skew_pandas)
print("Kurtosis pandas:", kurt_pandas)
print("Skewness scipy:", skew_scipy)
print("Kurtosis scipy:", kurt_scipy)
```

```python
def resumen_estadistico(serie, decimales=4):
    media = serie.mean()
    desv_std = serie.std()
    return {
        "n": serie.count(),
        "media": round(media, decimales),
        "mediana": round(serie.median(), decimales),
        "desv_std": round(desv_std, decimales),
        "cv_pct": round(desv_std / media * 100, decimales),
    }

print(resumen_estadistico(df["nota"]))
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"

for columna in ["nota", "asistencia_pct"]:
    resumen = resumen_estadistico(df[columna])
    nivel = clasificar_dispersion(resumen["cv_pct"])
    print(columna, "->", resumen, "- Dispersion:", nivel)
```
