```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

datos = {
    "centro": [45, 18.2, 72],
    "sanjeronimo":   [28, 12.5, 58],
    "wanchaq":   [62, 25.4, 81],
    "sansebastian":    [35, 15, 65],
    "poroy":    [0, 9.8, 52],
}
df = pd.DataFrame(datos)
corr = df.corr()
print(corr.round(4))
```

```python
plt.figure(figsize=(6, 5))
sns.heatmap(
    corr,
    annot=True, fmt=".2f",
    cmap="coolwarm",
    center=0, vmin=-1, vmax=1,
    square=True
)
plt.title("Matriz de correlación")
plt.tight_layout()
plt.savefig("heatmap_seaborn.png", dpi=150)
plt.show()
```

```python
mascara = np.triu(np.ones_like(corr, dtype=bool), k=1)
print("Celdas visibles:", int((~mascara).sum()))

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr, mask=mascara,
    annot=True, fmt=".2f",
    cmap="coolwarm", center=0, vmin=-1, vmax=1,
    square=True
)
plt.title("Matriz de correlación (sin redundancia)")
plt.tight_layout()
plt.savefig("heatmap_seaborn_mascara.png", dpi=150)
plt.show()
```
