```python
import numpy as np
import matplotlib.pyplot as plt

fila = np.linspace(0, 255, 5)
gris = np.tile(fila, (5, 1)).astype(np.uint8)

print(gris)
print("shape:", gris.shape)
print("dtype:", gris.dtype)
print("mean:", gris.mean())
```

```python
plt.imshow(gris, cmap="gray", vmin=0, vmax=255)
plt.title("Degradado horizontal 5x5")
plt.colorbar(label="Intensidad")
plt.savefig("gris_tema6.png")
plt.show()
```

```python
recorte = gris[:, 3:5]
print(recorte)
```

```python
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = (255, 0, 0)
rgb[:, 2:4] = (0, 0, 255)

print(rgb.shape)
print("Pixel [0,0]:", rgb[0, 0])
print("Pixel [0,3]:", rgb[0, 3])
```

```python
canal_r = rgb[:, :, 0]
canal_g = rgb[:, :, 1]
canal_b = rgb[:, :, 2]

print("Canal R:\n", canal_r)
print("Canal G:\n", canal_g)
print("Canal B:\n", canal_b)
```

```python
gris_promedio = rgb.mean(axis=2)
gris_luminosidad = (0.299 * rgb[:, :, 0] + 0.587 * rgb[:, :, 1] + 0.114 * rgb[:, :, 2])

print("Gris promedio:\n", gris_promedio)
print("Gris luminosidad:\n", gris_luminosidad)
```

```python
umbral = 50
binaria = (gris_luminosidad >= umbral).astype(np.uint8) * 255

print("Umbral usado:", umbral)
print("Imagen binarizada:\n", binaria)
```
