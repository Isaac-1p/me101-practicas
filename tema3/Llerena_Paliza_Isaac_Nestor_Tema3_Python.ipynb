```python
import numpy as np

notas = np.array([11, 14, 20, 8, 16, 13])

print(notas.shape, notas.dtype, notas.ndim)
```

```python
promedio = notas.mean()
maximo = notas.max()
cantidad_aprobados = (notas >= 10.5).sum()

print("Promedio:", promedio)
print("Maximo:", maximo)
print("Aprobados:", cantidad_aprobados)
```

```python
import pandas as pd

datos = {
    "nombre": ["Camila", "Diego", "Valeria", "Jorge", "Luz"],
    "nota": [16, 9, 13, 18, 11],
    "asistencia_pct": [90, 85, 60, 75, 72],
}
df = pd.DataFrame(datos)
print(df)

filtrados = df[(df["nota"] >= 10.5) & (df["asistencia_pct"] >= 70)]
print()
print(filtrados)
```

```python
import pandas as pd

datos = {
    "nombre": ["Ana", "Luis", "Marco", "Rosa", "Iván"],
    "nota":   [16,     12,     18,      9,     14],
    "ciclo":  [3,       5,      3,      1,      2],
}
df = pd.DataFrame(datos)
print(df.shape)
print(df.dtypes)
df.head()
```

```python
print(df['nota'].mean())              # 13.8
print(df.loc[0])                      # fila por etiqueta
print(df.iloc[0])                     # fila por posicion
aprobados = df.loc[df['nota'] >= 10.5]
print(aprobados[['nombre', 'nota']])
```
