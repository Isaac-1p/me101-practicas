```python
import pandas as pd, numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# --- Parte C ---
df = pd.DataFrame({
    'PM10': [45.0, 28.0, 62.0, 35.0, np.nan],
    'NO2':  [18.2, 12.5, 25.4, 15.0, 9.8],
    'Ruido':[72, 58, 81, 65, 52]},
    index=['Centro', 'San Jerónimo', 'Wanchaq', 'San Sebastián', 'Poroy'])
df_clean = df.fillna(df.median())
df_z = pd.DataFrame(StandardScaler().fit_transform(df_clean),
                    index=df_clean.index, columns=df_clean.columns)
print(df_z.round(2))

plt.figure(figsize=(7, 5))
sns.heatmap(df_z, cmap='coolwarm', center=0, annot=True, fmt='.2f',
            linewidths=0.5, linecolor='gray', cbar_kws={'label': 'Puntaje Z'})
plt.title('Indicadores de calidad ambiental (Z-score)', fontweight='bold')
plt.tight_layout()
plt.savefig('parteC_heatmap_python.png', dpi=300, bbox_inches='tight')
plt.show()

# --- Parte D ---
g = sns.clustermap(df_z, metric='euclidean', method='ward', cmap='coolwarm',
                   center=0, annot=True, fmt='.2f', linewidths=0.5,
                   figsize=(8, 7), dendrogram_ratio=(0.2, 0.2))
g.savefig('parteD_clustermap_python.png', dpi=300, bbox_inches='tight')
plt.show()
```
